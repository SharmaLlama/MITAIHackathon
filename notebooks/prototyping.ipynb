```python
import os
import shutil

# Define source folders and target folder
source_folders = ["path/to/train", "path/to/test", "path/to/val"]
target_folder = "path/to/combined"

# Create the target folder if it doesn't exist
os.makedirs(target_folder, exist_ok=True)

# Iterate through each source folder
for source_folder in source_folders:
    for filename in os.listdir(source_folder):
        source_file = os.path.join(source_folder, filename)
        target_file = os.path.join(target_folder, filename)
        
        # Check if the file already exists in the target folder
        if os.path.exists(target_file):
            # Rename the file to avoid conflicts
            base, ext = os.path.splitext(filename)
            counter = 1
            while os.path.exists(target_file):
                target_file = os.path.join(target_folder, f"{base}_{counter}{ext}")
                counter += 1
        
        # Copy the file to the target folder
        shutil.copy2(source_file, target_file)

print("Datasets combined successfully!")
```
